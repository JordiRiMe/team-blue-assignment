 team.blue is a subscription-based business selling services that help SMBs across Europe setup and run online businesses. Given that the revenue of the team.blue is subscription-based, one of the most important KPIs of the company is churn rate. That is the rate at which customers do not renew their subscription and are no longer a customer anymore. 

Your task in this assignment is to analyse customer invoive data and build a simple machine learning model to predict which customers are most likely to churn in their first year as a customer. This model can be used by the business to forecast the revenue impact of churning customers and to understand the drivers of churn. 

You have two datasets, one is for training and one is the validation set with no targets. We would like you to train a simple machine learning model (no deep learning, but you may use an ensemble should you want to) using the training set and give us predictions on the validation set. The deliverables are a jupyter notebook so that we can see your process and a csv with two columns ["FK_DWH_CUSTOMERS", "Predictions"], "Predictions" should be the predictions your model generated on the validation set; note predictions are not probabilities but a binary classification. 

The goal of the model is to predict which customers will be "noise" in 12 months (True or False). These are the customers that are expected to churn. The data you are provided has customer invoice data for the first 2 months of a customers lifespan with the company and has the following columns: 

1) PK_INVOICES = identifier code for a customer's invoice, multiple products can share the same invoice number
2) FK_DATE_INVOICE = The actual date of checkout YYYYMMDD as an int
3) CUSTOMER_START_DATE = The first day of the month that the customer made their first ever checkout.
4) TENURE = how many months since the customer became a customer (0 is first month, 1 is second month)
5) FLG_DWH_CUSTOMER_IS_BUSINESS = is the customer a business or private individual? (1 = business)
6) FK_DWH_CUSTOMERS = unique customer ID; note that most customers are present on more than 1 row
7) FK_SUBSCRIPTIONS = unique ID for a single product's subscription; unique to customer and product. a customer with two products will have    two unique FK_SUBSCRIPTIONS, but those unique identifiers will stay the same from month to month.
8) PRODUCT_GROUP = what the product is
9) PRODUCT_SUBCATEGORY = a less granular view of what the product is; there are fewer unique values in this column than in PRODUCT_GROUP
10) mrr = monthly recurring revenue; how much is paid for a subscription/product in that month by that customer.
11) ORIGIN_SUBGROUP = how was that customer acquired?
12) ORIGIN_GROUP = same as SUBGROUP, but less granular
13) noise = this is your target (y).

We would like to see your process, some sort of accuracy measure of your model (done on training data), some explanation of the model, and anything else that you might think is of interest.

While accuracy is very important, we are much more interested in your process and thinking than having a model that ekes out every last drop of possible accuracy.
